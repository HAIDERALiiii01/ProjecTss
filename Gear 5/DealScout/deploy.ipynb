# Setting Up Modal Tokens

**IMPORTANT — Please follow these instructions before deploying the fine-tuned model to Modal.**

First, visit [Modal](https://modal.com?utm_source=chatgpt.com) and create an account.

Once signed in:

1. Click your **Avatar** in the top-right corner.
2. Select **Settings**.
3. Open **API Tokens** from the left sidebar.
4. Click **New Token**.
5. Copy your **Token ID** and **Token Secret**.

Add both credentials to your project's `.env` file:

```env
MODAL_TOKEN_ID=ak-...
MODAL_TOKEN_SECRET=as-...
```

Make sure the `.env` file is in the root directory of your project.

Then load the environment variables in Python:

```python
from dotenv import load_dotenv

load_dotenv(override=True)
```

Your Modal credentials are now available to the deployment code, allowing you to deploy and access your fine-tuned pricing model on Modal.

> **Security:** Never commit your `.env` file or Modal tokens to GitHub. Add `.env` to your `.gitignore`.

# Setting Up the Hugging Face Secret in Modal

Before deploying the fine-tuned model, we need to give Modal access to your Hugging Face account.

Modal secrets have three parts:

- **Secret Name:** `huggingface-secret`
- **Key:** `HF_TOKEN`
- **Value:** Your Hugging Face token (`hf_...`)

### Steps

1. Go to [Modal](https://modal.com?utm_source=chatgpt.com) and sign in.
2. Open your **Dashboard**.
3. Navigate to **Secrets**.
4. Click **Create New Secret** and select **Hugging Face**.
5. Set the secret name to exactly:

```text
huggingface-secret
```

6. Set the key to:

```text
HF_TOKEN
```

7. Enter your actual Hugging Face token as the value:

```text
hf_...
```

8. Click **Done**.

The secret is now securely stored in Modal and can be accessed by the deployment code when downloading or using the fine-tuned model from Hugging Face.

> **Security:** Never put your Hugging Face token directly in your Python code or commit it to GitHub.
